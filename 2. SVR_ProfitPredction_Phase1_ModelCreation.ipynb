```python
# Importing Libraries
import pandas as pd 
import matplotlib.pyplot as plt
dataset=pd.read_csv("50_Startups.csv")
dataset=pd.get_dummies(dataset,drop_first=True)
dataset=dataset.astype(int)
independent=dataset[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
dependent=dataset[['Profit']]

#Standardization both input & Output
#Standardization for inputs
from sklearn.preprocessing import StandardScaler
sc_X=StandardScaler()
independent=sc_X.fit_transform(independent)

#Standardization for outputs
from sklearn.preprocessing import StandardScaler
sc_Y=StandardScaler()
dependent=sc_Y.fit_transform(dependent)
```

```python
# SVR Model Creation Using GridSearch Method
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
param_grid={'kernel':['linear', 'poly', 'rbf', 'sigmoid'], 'C':[0.1,1,10,100,1000], 'gamma':['scale','auto']}
grid=GridSearchCV(SVR(),param_grid,refit=True,verbose=3,n_jobs=-1)
grid=grid.fit(independent,dependent)
```

```python
re=grid.cv_results_
```

```python
table=pd.DataFrame.from_dict(re)
table
```

```python
# best Parameter
print("The best parameter is {}".format(grid.best_params_))
```

```python
import pickle
saved_model=pickle.dump(grid,open("Final_Model_SVR_ProfitPrediction.sav",'wb'))
loaded_model=pickle.load(open("Final_Model_SVR_ProfitPrediction.sav",'rb'))
```

```python
RnD_Spend=int(input("Enter the R&D spend value:"))
Admin=int(input("Enter the Administration :"))
MarketingSpend=int(input("Enter the Marketing Spend:"))
st_florida=int(input("Enter '1' if the state is florida else '0':"))
st_newYork=int(input("Enter '1' if the state is New York else '0':"))
```

```python
pre_input=sc_X.transform([[RnD_Spend,Admin,MarketingSpend,st_florida,st_newYork]])
```

```python
pre_input
```

```python
result=loaded_model.predict(pre_input)
pre_output=sc_Y.inverse_transform([result])
```

```python
print("The Predicted Profit Value is:",pre_output)
```
